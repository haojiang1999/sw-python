## analyse data from multiple data

```python
import glob

file_pattern =  "data/inflammation-*.csv"
all_files    = sorted(glob.glob(file_pattern))

data_arrays  = [];

for file in all_files:
    data=np.loadtxt(file, delimiter=",")
    data_arrays.append(data)
    print("Reading data from",file)
```

```python
len(data_arrays)
data_arrays[-1]
```

```python
for datasets in data_arrays:
    # mean/max/min inflammation per day per file
    mean_perday = np.mean(datasets, axis = 0)
    max_perday  = np.max(datasets, axis = 0)
    min_perday  = np.min(datasets, axis = 0)

    # create the figure of three subplots
    fig = plt.figure(figsize=(10,3))

    axes1 = fig.add_subplot(1,3,1)
    axes2 = fig.add_subplot(1,3,2)
    axes3 = fig.add_subplot(1,3,3)

    axes1.plot(mean_perday)
    axes2.plot(max_perday)
    axes3.plot(min_perday)

    axes1.set_ylabel("Mean")
    axes2.set_ylabel("Max")
    axes3.set_ylabel("Min")

    for ax in [axes1, axes2, axes3]:
        ax.set_xlabel("Day")
    
```

# Modular code / Documentation

```python
def reading_data(filename):
    """ Reading numerical data from a csv filec"""
    return np.loadtxt(filename, delimiter=",")

def analyze_data(data):
    """ Calculate the mean, maximum, minmum, and std of each colum """
    return np.mean(data, axis=0), np.max(data,axis=0), np.min(data,axis=0), np.std(data,axis=0)

def visualize_data(analyze_data):
    """ Plot the mean, maximum, and minimum values in three subplots """
    fig = plt.figure(figsize=(10,3))
    
    axes1 = fig.add_subplot(1,3,1)
    axes2 = fig.add_subplot(1,3,2)
    axes3 = fig.add_subplot(1,3,3)

    axes1.plot(analyze_data[0])
    axes2.plot(analyze_data[1])
    axes3.plot(analyze_data[2])

    for ax in [axes1, axes2, axes3]:
        ax.set_xlabel("Day")
    
    axes1.set_ylabel("Average")
    axes2.set_ylabel("Max")
    axes3.set_ylabel("Min")
    return fig
```

```python
for file in all_files[0:1]:
    data=reading_data(file)
    meanval, maxval, minval,stdval=analyze_data(data)
    fig = visualize_data([meanval,maxval,minval])
    fig.suptitle(file[5:-4], fontsize=16)
```
