## Exercise: Example of using and and or 

 

Write a program that prints whether a number is positive and even, positive and odd, zero or negative. 
Hint `num % X == 0` is true if `num` is divisible by `X`.

`````python
num = int(input("Enter a number: "))

if num > 0:
    if num % 2 == 0:
        print("Positive and even")
    else:
        print("Positive and odd")
elif num == 0:
    print("Zero")
else:
    print("Negative")
`````

## Exercise: What is truth? reflection 

True and False booleans are not the only values in Python that are true and false. In fact, any value can be used in an if or elif. 



`````python
if '':
    print('empty string is true')
if 'word':
    print('word is true')
if []:
    print('empty list is true')
if [1, 2, 3]:
    print('non-empty list is true')
if 0:
    print('zero is true')
if 1:
    print('one is true')
`````

`````python


import numpy as np
import matplotlib.pyplot as plt


def analyze_data(data):
    """ Calculate the mean, maximum, minmum, and std of each colum
    
    Parameters
    ----------
    data: numpy.ndarray
      filedata from csv file

    Returns:
    ----------
    maxval : numpy.ndarray
       maximum of each colum
    minval : numpy.ndarray
       minimum of each colum
    stdval : numpy.ndarray
       standard deviation of each colum
    meanval: numpy.ndarray
       mean of each colum
    
    """

    maxval, minval, stdval,meanval = np.amax(data), np.amin(data), np.std(data),np.mean(data)
    return maxval, minval, stdval,meanval

def visualize_data(data):
    """ Plot the mean, maximum, and minimum values in three subplots 
    
    Parameters
    ----------
    data: numpy.ndarray
      list includes meanvalue, maxvalue, and minvalue

    Returns:
    ----------
    fig : matplotlib.pyplot.figure
      plot including three subplots
    
    """

    fig=plt.figure(figsize=(15.0, 3.0))
    axes1=fig.add_subplot(1, 3, 1)
    axes2=fig.add_subplot(1, 3, 2)
    axes3=fig.add_subplot(1, 3, 3)

    axes1.plot(np.mean(data, axis=0))
    axes2.plot(np.amax(data, axis=0))
    axes3.plot(np.amin(data, axis=0))

    for ax in [axes1,axes2,axes3]:
        ax.set_xlabel('day')

    axes1.set_ylabel('mean')
    axes2.set_ylabel('max')
    axes3.set_ylabel('min')

    return fig
`````
